# Day 4 — Performance Analytics
This notebook runs the `scripts/performance_analytics.py` script, then loads and displays the key CSV outputs and charts produced by the script for quick interactive review.

```python
# Run the analytics script (safe to re-run)
import subprocess, sys, pathlib
ROOT = pathlib.Path('..').resolve() if pathlib.Path.cwd().name == 'notebooks' else pathlib.Path.cwd()
script = ROOT / 'scripts' / 'performance_analytics.py'
subprocess.run([sys.executable, str(script)], check=True)
```

```python
# Display top rows of the fund scorecard and alpha/beta
import pandas as pd, pathlib, IPython.display as disp
OUT = (pathlib.Path('..') / 'reports' / 'performance').resolve() if pathlib.Path.cwd().name == 'notebooks' else pathlib.Path('reports') / 'performance'
score = pd.read_csv(OUT / 'fund_scorecard.csv')
alpha = pd.read_csv(OUT / 'alpha_beta.csv')
disp.display(score.sort_values('score', ascending=False).head(10))
disp.display(alpha.head(10))
```

```python
# Show benchmark comparison image and validation plots
from IPython.display import Image, display
display(Image(str(OUT / 'benchmark_comparison_top5.png')) )
display(Image(str(OUT / 'daily_returns_hist.png')) )
display(Image(str(OUT / 'sharpe_sortino_hist.png')) )
display(Image(str(OUT / 'alpha_beta_scatter.png')) )
```
